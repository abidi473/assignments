```python
import pandas as pd
# Dataset_1
ecom_data = pd.read_csv(r"C:\Users\Abid\Downloads\pakistani_ecommerce_dataset.csv")
ecom_data
```

```python
ecom_data.groupby("product_category").size()
```

```python
# Answer 1: The most popular category is home decor.
ecom_data["order_value"].mean().round(2)
```

```python
# Answer 2: 2832.15 is the average order value.
ecom_data.groupby("city").size()
```

```python
# Answer 3: Karachi has the highest number of orders.
ecom_data.groupby("payment_method").size()
```

```python
# Answer 4: Cash on delivery is the most used payment method.
import matplotlib.pyplot as plt
avg_order_by_city = ecom_data.groupby("city")["order_value"].mean()

avg_order_by_city.plot(kind="bar")

plt.title("Average Order Value by City")
plt.xlabel("City")
plt.ylabel("Average Order Value")
plt.xticks(rotation=45)
plt.show()
# Answer 5:
```
