```python
import pandas as pd
```

```python
clean = pd.read_csv(r"C:\assignments\week 2\teaching_prices_clean.csv")
```

```python
clean.dtypes
```

```python
clean["price"] = clean["price"].str.replace("9999", "np.nan")
```

```python
clean["price"]=pd.to_numeric(clean["price"].str.replace (",", ""), errors = "coerce")
```

```python
clean
```

```python
clean.dtypes
```

```python
clean["item"]= clean["item"].str.title()
```

```python
clean.isna().sum()
```

```python
clean["price"] = clean["price"].fillna(clean["price"].median())
```

```python
clean.isna().sum()
```

```python
clean.groupby("item")["price"].agg(["mean", "median"])
```

```python
clean['date'] = pd.to_datetime(clean['date'])
clean = clean.sort_values('date')
result = clean.groupby('item')['price'].agg(['first', 'last'])
result['percentage_change'] = (
    (result['last'] - result['first']) / result['first']
) * 100
print(result.round(2))
```
